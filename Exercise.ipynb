#### Exercise 2

Use the model of Fig. 2.18, but distribute the solar heating such that $0.3\sigma T_e^4$; is absorbed in each of the two atmospheric layers and the remaining $0.4\sigma T_e^4$ is absorbed at the surface. Calculate the new radiative equilibrium temperature profile. How does it differ from the case where all of the solar heating is applied at the surface.

**Solution :** 

 The energy balance of the system will change as the solar heating will be distributed in the atmosphere (which becomes non-transparent to SW).

 - At the top of the atmosphere (TOA):

 $$\dfrac{S_0}{4}(1-\alpha) = \sigma T_1^4$$

 - In layer 1,

  $$0.3 \times \dfrac{S_0}{4}(1-\alpha)  + \sigma T_2^4  = 2 \sigma T_1^4$$

- In layer 2,

  $$0.3 \times \dfrac{S_0}{4}(1-\alpha)  + \sigma T_1^4  + \sigma T_S^4  = 2 \sigma T_2^4 $$

- And at the surface, 

 $$0.4 \times \dfrac{S_0}{4}(1-\alpha)  + \sigma T_2^4  = \sigma T_S^4$$


 The system is solved analytically by direct substitution using the given equations.

From layer 1:  
$$\sigma T_1^4 = \dfrac{S_0}{4}(1-\alpha)$$

Substitute into layer 2 equation:  
$$ 0.3 \times \dfrac{S_0}{4}(1-\alpha) + \sigma T_2^4 = 2 \sigma T_1^4 = 2 \times \dfrac{S_0}{4}(1-\alpha)$$
$$ \sigma T_2^4 = 2 \times \dfrac{S_0}{4}(1-\alpha) - 0.3 \times \dfrac{S_0}{4}(1-\alpha) = 1.7 \times \dfrac{S_0}{4}(1-\alpha)$$

Substitute into surface equation: 
$$0.4 \times \dfrac{S_0}{4}(1-\alpha) + \sigma T_2^4 = \sigma T_S^4$$

$$\sigma T_S^4 = 0.4 \times \dfrac{S_0}{4}(1-\alpha) + 1.7 \times \dfrac{S_0}{4}(1-\alpha) = 2.1 \times \dfrac{S_0}{4}(1-\alpha)$$

**Thus,**

$$T_1 = \left[ \dfrac{S_0}{4\sigma}(1-\alpha) \right]^{1/4}$$

$$T_2 = \left[ \dfrac{1.7 S_0}{4\sigma}(1-\alpha) \right]^{1/4}$$

$$T_S = \left[ \dfrac{2.1 S_0}{4\sigma}(1-\alpha) \right]^{1/4}$$




```python
# Compute the profile

S0 = 1370                                       #W/m^2
sigma = 5.67e-8                                 #W/m^2K^4
alpha = 0.3

T1 = (S0*(1-alpha)/(4*sigma))**(1/4)
T2 = (1.7*S0*(1-alpha)/(4*sigma))**(1/4)
TS = (2.1*S0*(1-alpha)/(4*sigma))**(1/4)

print('T1 = ', T1, '\n' 'T2 = ', T2, '\n' 'TS =', TS)
```

**The temperature profile suggests more stable (lower lapse rate) compared to the generic SW transparent atmosphere. The fact that $60\%$ of the solar heating is distributed to the layers of the atmosphere warms the atmosphere above the surface, which will stabilize the temperature profile. Therefore,  although the surface still receives energy flux from the warm overlying layer, the energetic solar heating at the surface gets smaller so that the net effect is cooler surface.**

- Yuma is located close to the border of Az and CA, which has a desert like dry climate at a low altitude around $100\;m$, and also relatively higher albedo.
- Flagstaff, however, is located in the interior highlands of Az with altitude around $2000\;m$, has more vegetation cover and therefore slightly lower albedo.

The analysis is based on the following: $$ R_s =  SW(down) - SW(up) + LW(down) - LW(up)$$

During summer, the incoming SW ($SW(down)$) radiation peaks (say the are in the same band of latitudes). But given that Flagstaff is at a higher altitude, then the tempearture is expected to be lower (lapse rate) compared to Yuma, which is actually hot. This implies that the LW emission ($LW(up) = \sigma T^4$) in Flagstaff is expected to be lower, which makes the net LW less negative. At the same time, the LW emission is large in Yuma as the temperature is high. Therefore, there is a larger net LW cooling in Yuma compared to Flagstaf. Moreover, the slightly higher albedo in Yuma reduces the surface net SW radiation, which also contribute to contrast with Flagstaff.
